```python
import pandas as pd

base_path = r"C:\Users\DELL\Desktop\DM\data"

files = [
    "begin_inventory.csv",
    "end_inventory.csv",
    "purchase_prices.csv",
    "purchases.csv",
    "sales.csv",
    "vendor_invoice.csv"
]

for filename in files:

    path = f"{base_path}\\{filename}"
    df = pd.read_csv(path)

    print("\n" + "=" * 100)
    print(f"FILE: {filename}")
    print("=" * 100)

    for col in df.columns:
        print(
            f"{col:<25} "
            f"dtype={str(df[col].dtype):<10} "
            f"nulls={df[col].isna().sum():<8} "
            f"sample={df[col].dropna().head(3).tolist()}"
        )
```
